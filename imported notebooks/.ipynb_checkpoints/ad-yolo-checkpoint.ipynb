```python
import time
try:
    from pynvml import *
    nvmlInit()
    deviceCount = nvmlDeviceGetCount()
    if deviceCount > 0:
        handle = nvmlDeviceGetHandleByIndex(0) # Test with the first GPU
        deviceName = nvmlDeviceGetName(handle)
        print(f"Attempting to get energy consumption for {deviceName}") # Removed .decode() again
        # The energy consumption is cumulative since the last driver load
        # To see a change, you might need to run a GPU workload
        initial_energy = nvmlDeviceGetTotalEnergyConsumption(handle)
        print(f"Initial energy consumption (mJ): {initial_energy}")
        time.sleep(5) # Wait for a few seconds
        current_energy = nvmlDeviceGetTotalEnergyConsumption(handle)
        print(f"Current energy consumption (mJ) after 5 seconds: {current_energy}")
        if current_energy > initial_energy:
            print("Energy consumption is being tracked.")
        else:
            print("Energy consumption does not appear to be changing or is not being tracked.")

    nvmlShutdown()
except NVMLError as error:
    print(f"Error accessing NVML energy data: {error}")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

```python
# Here we have a custom gpu power draw calculator, due to pynvml not being able to detect the gpu power draw of Tesla P100s
```

```python
import subprocess
import time
import threading
import queue
import sys
import io # To capture output if needed

# Event to signal the monitoring thread to stop
stop_monitoring = threading.Event()
# List to hold power draw values for averaging
collected_power_draws = []
# Lock to ensure thread-safe access to collected_power_draws
power_draws_lock = threading.Lock()

def monitor_gpu_power(interval=5):
    """
    Monitors GPU power draw using nvidia-smi at specified intervals
    and stores the readings.
    """
    print(f"Starting GPU power monitoring every {interval} seconds...")
    while not stop_monitoring.is_set():
        try:
            # Query power draw for all GPUs
            result = subprocess.run(
                ["nvidia-smi", "--query-gpu=power.draw", "--format=csv,noheader,nounits"],
                capture_output=True,
                text=True,
                check=True
            )
            # Split the output into individual GPU readings
            readings_str = result.stdout.strip().split('\n')
            current_readings = [float(r) for r in readings_str if r] # Convert to float, handle empty lines

            # Store readings with timestamp in a thread-safe manner
            with power_draws_lock:
                 collected_power_draws.append(current_readings)

        except FileNotFoundError:
            print("Error: nvidia-smi not found. Make sure NVIDIA drivers are installed and in your PATH.")
            stop_monitoring.set() # Signal to stop if nvidia-smi is not found
        except subprocess.CalledProcessError as e:
            print(f"Error querying nvidia-smi: {e}")
        except Exception as e:
            print(f"An unexpected error occurred during monitoring: {e}")

        # Wait for the next interval or stop signal
        stop_monitoring.wait(interval)
    print("GPU power monitoring stopped.")

# You will start the thread in the next cell
monitor_thread = None
```

```python
# In this snippet, we can replace the mock training with the actual training for our yolo-model.
```

```python
# Reset the stop event and clear previous readings for a new run
from ultralytics import YOLO
stop_monitoring.clear()
collected_power_draws = [] # Clear previous data

# Start the monitoring thread
monitor_thread = threading.Thread(target=monitor_gpu_power, args=(5,)) # Monitor every 5 seconds
monitor_thread.daemon = True # Allow the notebook to exit even if this thread is running
monitor_thread.start()

# --- Your Model Training Code Starts Here ---
# Place all your training setup, data loading, model definition,
# and training loops in this cell or subsequent cells.

print("Starting model training...")

# Example training placeholder:
model = YOLO('yolo11s.pt')  # Start with the small model for better quality
model.train(data=f'{base_folder}/data.yaml', epochs=35, imgsz=1024, device=[1], batch=32,
    degrees=10,         # random rotation (degrees)
    translate=0.1,      # random translation
    scale=0.5,          # random scaling
    shear=3,            # random shear
    perspective=0.0,    # random perspective
    flipud=0.0,         # vertical flip probability
    fliplr=0.5,         # horizontal flip probability
    mosaic=0.5,         # mosaic augmentation probability
    mixup=0.2,          # mixup augmentation probability
    hsv_h=0.0,          # disable hue augmentation (important for your data)
    hsv_s=0.0,          # disable saturation augmentation
    hsv_v=0.3,
)
print("Model training finished.")

# --- Your Model Training Code Ends Here ---

# Signal the monitoring thread to stop
stop_monitoring.set()

# Wait for the monitoring thread to finish (optional, but good practice)
if monitor_thread and monitor_thread.is_alive():
    print("Waiting for monitoring thread to finish...")
    monitor_thread.join(timeout=10) # Give it some time to stop

print("\nCalculating average GPU power draw...")

# Calculate and print the average power draw
if collected_power_draws:
    # collected_power_draws is a list of lists, where each inner list
    # contains the power draw for each GPU at a specific timestamp.
    # We need to average across all readings and for each GPU.

    num_gpus = len(collected_power_draws[0]) if collected_power_draws and collected_power_draws[0] else 0
    if num_gpus > 0:
        average_power_per_gpu = [0.0] * num_gpus
        total_readings_per_gpu = [0] * num_gpus

        with power_draws_lock: # Acquire lock before accessing collected_power_draws
            for readings_list in collected_power_draws:
                for i in range(num_gpus):
                    if i < len(readings_list):
                         average_power_per_gpu[i] += readings_list[i]
                         total_readings_per_gpu[i] += 1

        print("\n--- Average GPU Power Draw ---")
        for i in range(num_gpus):
            if total_readings_per_gpu[i] > 0:
                avg_power = average_power_per_gpu[i] / total_readings_per_gpu[i]
                print(f"GPU {i}: {avg_power:.2f} W (averaged over {total_readings_per_gpu[i]} readings)")
            else:
                 print(f"GPU {i}: No readings collected for this GPU.")
        print("------------------------------")

    else:
        print("No GPU power readings collected or no GPUs detected in readings.")

else:
    print("No GPU power data was collected during the run.")
```
